## Q1. Campus Resource Tracker

**Sample data**

```python
resources = [
    [101, "Laptop", "Electronics", 8],
    [102, "Projector", "Electronics", 0],
    [103, "Whiteboard", "Classroom", 5],
    [104, "Microphone", "Audio", 3],
    [105, "Camera", "Electronics", 2]
]
```

**Task 1: Display all resources**

```python
def display_resources(resources):
    if len(resources) == 0:
        print("No resources to display.")
        return

    print(f"{'ID':<6}{'Name':<12}{'Category':<13}{'Quantity':>8}")
    print("-" * 39)
    for res in resources:
        res_id, name, category, quantity = res
        print(f"{res_id:<6}{name:<12}{category:<13}{quantity:>8}")


display_resources(resources)
```

**Task 2: Search by category**

```python
def search_by_category(resources, category):
    found = []
    for res in resources:
        if res[2].lower() == category.lower():
            found.append(res)

    if len(found) == 0:
        print(f"No resources found in category '{category}'.")
    else:
        print(f"Resources in category '{category}':")
        display_resources(found)
    return found


search_by_category(resources, "Electronics")
print()
search_by_category(resources, "Furniture")
```

**Task 3: Find unavailable resources**

```python
def find_unavailable(resources):
    unavailable = []
    for res in resources:
        if res[3] == 0:
            unavailable.append(res)

    if len(unavailable) == 0:
        print("All resources are available.")
    else:
        print("Unavailable resources (quantity = 0):")
        for res in unavailable:
            print(f"  {res[0]} - {res[1]}")
    return unavailable


find_unavailable(resources)
```

**Testing and debugging**

```python
def find_by_id(resources, res_id):
    for res in resources:
        if res[0] == res_id:
            return res
    return None


def add_resource(resources, res_id, name, category, quantity):
    if find_by_id(resources, res_id) is not None:
        print(f"Error: resource ID {res_id} already exists.")
        return False
    if quantity < 0:
        print(f"Error: quantity cannot be negative ({quantity}) for '{name}'.")
        return False
    resources.append([res_id, name, category, quantity])
    print(f"Added resource {res_id} - {name}.")
    return True


# Test 1: unknown resource ID
print("Test 1: unknown resource ID")
result = find_by_id(resources, 999)
if result is None:
    print("Resource ID 999 not found.")

# Test 2: negative quantity
print("\nTest 2: negative quantity")
add_resource(resources, 106, "Speaker", "Audio", -4)

# Test 3: resource with zero units
print("\nTest 3: resource with zero units")
add_resource(resources, 107, "Tripod", "Electronics", 0)
find_unavailable(resources)

# Test 4: empty resource list
print("\nTest 4: empty resource list")
empty = []
display_resources(empty)
search_by_category(empty, "Electronics")
find_unavailable(empty)
```

## Q2. Student Marks Investigator

**Sample data**

```python
students = [
    [201, "Amina", [78, 85, 90]],
    [202, "Rahim", [65, 70, 68]],
    [203, "Nadia", [92, 88, 95]],
    [204, "Hasan", [55, 60, 65]],
    [205, "Tania", [80, 80, 80]]
]
```

**Task 1: Calculate average**

```python
def calculate_average(marks):
    if len(marks) == 0:
        return None
    for m in marks:
        if not isinstance(m, (int, float)) or m < 0 or m > 100:
            return None
    return sum(marks) / len(marks)


print(calculate_average([78, 85, 90]))
print(calculate_average([80, 80, 80]))
```

**Task 2: Classify performance**

```python
def classify_performance(average):
    if average is None:
        return "Invalid Data"
    if average >= 80:
        return "Excellent"
    elif average >= 70:
        return "Good"
    elif average >= 60:
        return "Satisfactory"
    else:
        return "Needs Improvement"


for avg in [84.3, 70, 60, 59.9]:
    print(avg, "->", classify_performance(avg))
```

**Task 3: Display report**

```python
def display_report(students):
    if len(students) == 0:
        print("No student records to display.")
        return

    print(f"{'ID':<6}{'Name':<10}{'Average':>8}   Performance")
    print("-" * 42)
    for student in students:
        sid, name, marks = student
        avg = calculate_average(marks)
        category = classify_performance(avg)
        if avg is None:
            if len(marks) == 0:
                reason = "no marks"
            else:
                reason = "invalid mark"
            print(f"{sid:<6}{name:<10}{'N/A':>8}   {category} ({reason})")
        else:
            print(f"{sid:<6}{name:<10}{avg:>8.2f}   {category}")


display_report(students)
```

**Testing and debugging**

```python
test_students = [
    [301, "Equal", [75, 75, 75]],      # equal marks in all assessments
    [302, "Tie1", [70, 80, 90]],       # average 80
    [303, "Tie2", [85, 75, 80]],       # average 80 (tie with Tie1)
    [304, "Invalid", [90, 105, 80]],   # 105 is not a valid mark
    [305, "Negative", [-5, 60, 70]],   # -5 is not a valid mark
    [306, "NoMarks", []]               # no marks
]

display_report(test_students)


def find_ties(students):
    groups = {}
    for sid, name, marks in students:
        avg = calculate_average(marks)
        if avg is not None:
            groups.setdefault(round(avg, 2), []).append(name)

    found = False
    for avg, names in groups.items():
        if len(names) > 1:
            print(f"Tied average {avg:.2f}: {', '.join(names)}")
            found = True
    if not found:
        print("No tied averages.")


print()
find_ties(test_students)
print()
print("Ties in the sample data:")
find_ties(students)
```
