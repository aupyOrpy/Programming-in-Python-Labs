## Q1. Swap two variables

```python
a = input("Enter value of a: ")
b = input("Enter value of b: ")

print("Before swap: a =", a, ", b =", b)

# Method 1: Pythonic swap
a, b = b, a
print("After swap (a, b = b, a): a =", a, ", b =", b)

# Method 2: using a temporary variable (swaps them back)
temp = a
a = b
b = temp
print("After swap (temp variable): a =", a, ", b =", b)
```

## Q2. Distance between two points

```python
import math

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

d = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print(f"Distance between ({x1}, {y1}) and ({x2}, {y2}) = {d:.2f}")
```

## Q3. Even or odd number

```python
n = int(input("Enter a number: "))

if n % 2 == 0:
    print(n, "is an even number")
else:
    print(n, "is an odd number")
```

## Q4. Vowel or not vowel

```python
ch = input("Enter a character: ")

if len(ch) != 1 or not ch.isalpha():
    print("Invalid input: please enter a single letter")
elif ch.lower() in "aeiou":
    print(ch, "is a vowel")
else:
    print(ch, "is not a vowel")
```

## Q5. Marks to grade (AIUB grading)

```python
marks = float(input("Enter marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
else:
    if marks >= 90:
        grade, point = "A+", 4.00
    elif marks >= 85:
        grade, point = "A", 3.75
    elif marks >= 80:
        grade, point = "B+", 3.50
    elif marks >= 75:
        grade, point = "B", 3.25
    elif marks >= 70:
        grade, point = "C+", 3.00
    elif marks >= 65:
        grade, point = "C", 2.75
    elif marks >= 60:
        grade, point = "D+", 2.50
    elif marks >= 50:
        grade, point = "D", 2.25
    else:
        grade, point = "F", 0.00

    print(f"Grade: {grade}  (Grade point: {point:.2f})")
```
