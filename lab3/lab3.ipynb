## Q1. Write a Python program to swap the values of two variables.

### Solution

Python can swap two variables in one line with tuple assignment. The classic method with a temporary variable is also shown.

```python
a = input("Enter value of a: ")
b = input("Enter value of b: ")

print("Before swap: a =", a, ", b =", b)

# Method 1: Pythonic swap
a, b = b, a
print("After swap (a, b = b, a): a =", a, ", b =", b)

# Method 2: using a temporary variable (swaps them back)
temp = a
a = b
b = temp
print("After swap (temp variable): a =", a, ", b =", b)
```

## Q2. Write a program to compute the distance between two points, taking input from the user.

Distance formula: $d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$

```python
import math

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

d = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print(f"Distance between ({x1}, {y1}) and ({x2}, {y2}) = {d:.2f}")
```

## Q3. Write a program to check whether the given number is an even or odd number.

### Solution

A number is even when the remainder after dividing by 2 is 0 (`n % 2 == 0`), otherwise it is odd.

```python
n = int(input("Enter a number: "))

if n % 2 == 0:
    print(n, "is an even number")
else:
    print(n, "is an odd number")
```

## Q4. Take a character as input and determine whether it is a vowel. Print an appropriate message (vowel/not vowel).

### Solution

The input is converted to lowercase so both `A` and `a` work. The `in` membership operator checks if it is one of `a, e, i, o, u`. Inputs that are not a single letter are rejected.

```python
ch = input("Enter a character: ")

if len(ch) != 1 or not ch.isalpha():
    print("Invalid input: please enter a single letter")
elif ch.lower() in "aeiou":
    print(ch, "is a vowel")
else:
    print(ch, "is not a vowel")
```

## Q5. Write a program to convert a student's marks of a course into a grade. Use the grading criteria of our university.

### Solution

AIUB grading criteria:

| Marks | Letter Grade | Grade Point |
|---|---|---|
| 90 – 100 | A+ | 4.00 |
| 85 – 89 | A | 3.75 |
| 80 – 84 | B+ | 3.50 |
| 75 – 79 | B | 3.25 |
| 70 – 74 | C+ | 3.00 |
| 65 – 69 | C | 2.75 |
| 60 – 64 | D+ | 2.50 |
| 50 – 59 | D | 2.25 |
| below 50 | F | 0.00 |

```python
marks = float(input("Enter marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
else:
    if marks >= 90:
        grade, point = "A+", 4.00
    elif marks >= 85:
        grade, point = "A", 3.75
    elif marks >= 80:
        grade, point = "B+", 3.50
    elif marks >= 75:
        grade, point = "B", 3.25
    elif marks >= 70:
        grade, point = "C+", 3.00
    elif marks >= 65:
        grade, point = "C", 2.75
    elif marks >= 60:
        grade, point = "D+", 2.50
    elif marks >= 50:
        grade, point = "D", 2.25
    else:
        grade, point = "F", 0.00

    print(f"Grade: {grade}  (Grade point: {point:.2f})")
```
