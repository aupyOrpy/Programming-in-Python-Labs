## Q1

```python
fruits = ["apple", "banana", "cherry", "mango"]

print("Looping over a list:")
for fruit in fruits:
    print(fruit)

print("\nLooping over a string:")
for ch in "Python":
    print(ch, end=" ")
print()

print("\nLooping with index using enumerate():")
for ind, val in enumerate(fruits):
    print(ind, val)

print("\nLooping over a range (squares of 1 to 5):")
for i in range(1, 6):
    print(i, "squared is", i ** 2)
```

## Q2

```python
n = int(input("Enter a number: "))

if n < 0:
    print("Please enter a non-negative number")
else:
    while n >= 0:
        print(n)
        n -= 1
    print("Countdown finished!")
```

## Q3

```python
base = float(input("Enter base: "))
exponent = int(input("Enter exponent (integer): "))

result = 1
for i in range(abs(exponent)):
    result *= base

if exponent < 0:
    result = 1 / result

print(f"{base} ^ {exponent} = {result}")
print("Using ** operator:", base ** exponent)
```

## Q4

```python
total = 0
count = 0

for n in range(2, 1000):
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        total += n
        count += 1

print("Number of primes below 1000:", count)
print("Sum of all primes below 1000:", total)
```

## Q5

```python
N = int(input("Enter N: "))

a, b = 0, 1
print(f"Fibonacci numbers below {N}:")
while a < N:
    print(a, end=" ")
    a, b = b, a + b
print()
```
